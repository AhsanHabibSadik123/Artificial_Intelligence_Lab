```python
from PIL import Image
import os

# ==========================
# Configuration
# ==========================
INPUT_FOLDER = "/home/sadik/Desktop/qw/images/kathalpata"
OUTPUT_FOLDER = "/home/sadik/Desktop/qw/final_images/kathalpata"
    
SIZE = (256, 256)

# Supported image extensions
EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp")

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# Get image files
image_files = [
    f for f in os.listdir(INPUT_FOLDER)
    if f.lower().endswith(EXTENSIONS)
]

# Sort for consistent numbering
image_files.sort()

for idx, filename in enumerate(image_files, start=1):
    input_path = os.path.join(INPUT_FOLDER, filename)
    output_path = os.path.join(OUTPUT_FOLDER, f"image_{idx}.jpg")

    try:
        with Image.open(input_path) as img:
            # Convert to RGB (handles PNG with alpha, grayscale, etc.)
            img = img.convert("RGB")

            # Resize
            img = img.resize(SIZE, Image.Resampling.LANCZOS)

            # Save as JPG
            img.save(output_path, "JPEG", quality=95)

        print(f"Saved: {output_path}")

    except Exception as e:
        print(f"Failed to process {filename}: {e}")

print(f"\nDone! Processed {len(image_files)} images.")
```

```python
import os
import pandas as pd

# ==========================
# Dataset Configuration
# ==========================
ROOT_DIR = "/home/sadik/Desktop/qw/final_images"

# Class name : numeric label
CLASS_FOLDERS = {
    "aampata": 0,
    "kathalpata": 1
}

IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

rows = []

for class_name, label in CLASS_FOLDERS.items():
    folder_path = os.path.join(ROOT_DIR, class_name)

    # Sort image_1.jpg, image_2.jpg, ...
    image_files = sorted(
        [f for f in os.listdir(folder_path) if f.lower().endswith(IMAGE_EXTENSIONS)],
        key=lambda x: int(os.path.splitext(x)[0].split("_")[-1])
    )

    for image_name in image_files:
        rows.append({
            "image_path": os.path.join(class_name, image_name),
            "label": label,
            "class_name": class_name
        })

# Create DataFrame
df = pd.DataFrame(rows)

# Save CSV
csv_path = os.path.join(ROOT_DIR, "labels.csv")
df.to_csv(csv_path, index=False)

print(f"CSV saved to: {csv_path}")
print(f"Total images: {len(df)}")
print(df.head())
```
